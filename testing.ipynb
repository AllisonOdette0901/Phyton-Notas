## Applying test markers

Un marcadorde prueba es una etiqueta de una prueba en la biblioteca pytest, se usa para especificar el comportamiento de una prueba en particular.

- **Decorador:** es un patrón de diseño en python que pemrite a un usuario agrgear una nueva funcionalidad a un objeto existente sin modificar su estructura.
- **Test markers syntax:** agregamos el decorador que comienza con `@pytest.mark`

```python
import pytest

def get_length(word):
    return len(word)

# The test marker syntax
@pytest.mark.skip
def test_get_length():
    assert get_length("hello") == 5
    assert get_length("bye") == 3
```

### Skip and skipif markers
Hay marcadores que se pueden usar de inmediato:
- Uno es el marcador `@pytest.mark.skip` cualquier prueba marcada con este se omitirá durante la ejecución de pytest. Es útil si uno quiere esperar para ejecutar esa prueba.
- Otro es `@pytest.mark.skipif` cualquier prueba marcada con él se omitirá, pero sólo si la condición dada es verdadera.

Vamos a ver un ejemplo de cómo utilizar skipif al mismo fragmento, le pasamos 2*2 == 5 lo cual siempre es falso, la prueba no se imitirá y se ejecutará. Pero también podemos pasar las variables al skipif, por ejemplo, para verificar las versiones del módulo.

```python
import pytest

def get_length(word):
    return len(word)

# The test marker syntax
@pytest.mark.skipif('2 * 2 == 5')
def test_get_length():
    assert get_length("bye") == 3
```

### Xfail marker
- Usamos `@ptest.mark.xfail` cuando sabemos que va a fallar el test.

La prueba terminará como "exitosa" porque fallará exactamente como esperabamos.

```python
import pytest

def gen_sequence(n):
    return list(range(1,n+1))

# The test marker syntax
@pytest.mark.xfail
def test_gen_seq():
    assert gen_sequence(-1)
```

# Introduction a fixtures

- **Fixture(accesorio):** es un entorno que se puede utilizar para la ejecución de pruebas.
Por ejemplo, si queremos utiliza un conjutno de datos para las pruebas, el proceso de preparación de dichjo entorno se denomina configuración del dispositivo.
- **Fixture Setup**

Fixtures hepl:
- Aíslan la prueba de la preparación ambiental.
- Los accesorios hacen qu el código sea reutilizable especialmente si queremos usar el mismo código de preparación par diferentes pruebas.

```python
import pytest

#fixture decorator
# le hacemos saber a phyton que no es una función normal sino un elemento fijo
@pytest.fixture
#fixture for data visualization
def data():
  return [0,1,1,2,3,5,8,13,21]
def test_list(data):
  assert len(data) == 9
  assert 5 in data
  assert 21 in data
```

Pasos para usar fixtures:
1. Preparar software y tests
2. Encontrar "enviroment preparation"
3. Crear una fixture:
  - Declarar el `@pytest.fixture` decorator
  - Implementar la función fixture
4. Usar la fixture creada:
  - Pasamos el nombre de fixture en la función test
  - Corremos el tests.

```python
import pytest

@pytest.fixture
def prepare_data():
    return [i for i in range(10)]

def test_elements(prepare_data):
    assert 9 in prepare_data
    assert 10 not in prepare_data

# y corremos lo siguiente el consola
pytest run_with_fixtures.py
# run_with_fixtures es el nombre del archivo
```

## Chain Fixtures requets

Las solicitudes de dispositivos en cadena son una característica que permite que un dispositivo solicite otro dispositivo(fixture).

## Autouse argument
Es un argumento boolenao opcional de un dispositivo que se puede pasar al decorador de la fixture.

Cuando autouse sea igual a True, la función de fijación se ejecuta independientemente de si fue solicitada o no.

Lo usamos cuando tenemos qu aplicar ciertas preparaciones o modificaciones ambiental es para las pruebas al mismo tiempo. Por ejemplo, cuando todas las pruebas se ejecutan con los mismos datos, utilizando las mismas conexiones, el mismo entorno, o simplemente queremos monitorear el proceso.

La forma de usarlo es `@pytest.fixture(autouse=True)`

La razón princiapl para usarlo es reducir la cantidad de llamadas a dispositivos redundantes para simplificar nuestro código.


```python
import pytest

@pytest.fixture
def init_list():
    return []

# Declare the fixture with autouse
@pytest.fixture(autouse=True)
def add_numbers_to_list(init_list):
    init_list.extend([i for i in range(10)])

# Complete the tests
def test_elements(init_list):
    assert 5 in init_list
    assert 9 in init_list
```

## Fixtures Teardowns

La falta de teardowns (desmontajes) puede generar problemas importantes como:
- Pérdida de memoria
- Baja velocidad de ejecución y problemas de rendimiento,
- Resultados de pruebas no válidos y fallas en la canalización(pipeline).

- Definición: Es un proceso de limpieza de recursos que se asignaron o crearon durante la configuración de un entrono de prueba.
- Usage:
  - Podemos implementar un teardown ingresando la palabra clave "yield" en lugar de "return"

```python
import pytest

@pytest.fixture
def prepare_data():
    data = [i for i in range(10)]
    # Return the data with the special keyword
    yield data
    # Clear the data list
    data.clear()
    # Delete the data variable
    del data

def test_elements(prepare_data):
    assert 9 in prepare_data
    assert 10 not in prepare_data
```

# Pruebas unitarias

- **Unit:** son las partes ufncionales más pequeñas que se pueden probar.
Normalmente represnetan: funciones, métodos, pero a veces módulos o clases.

- **Unit testing:** es una prueba que verifica que una unidad de software funcione como se espera. Se espera que el fragmento de código más pequeño realmente funcione.

- **Test case:** es un conjunto de pruebas unitarias y salidas esperadas.

Es muy importante porque un caso d eprueba es en realidad un modelo para prueba futura.

Pasos:

1. Decidir las unidades que queremos probar.
2. Definir loscasos de prueba (the creative part):
  - Cuáles son los posibles resultados de la unidad?
  - Cómo se puede utilizar la unidad?
  - Cómo debería comportarse la unidad en estos casos?
3. Escribir código fácilmnete para cada paso de prueba.

```python
# función para suma de elementos
def sum_of_list(values) -> int:
    return sum(values)
```

Test cases:
1. Input es una lista de números - retorna la suma `sum`
2. Considerar casos no triviales, como si la entrada es vacía o una lista contiene un número.


```python
def test_regular():
    assert sum_of_list([1,2,3]) == 6
    assert sum_of_list([1,2,3,4]) == 10

# test case 2: empty list
def test_empty():
    assert sum_of_list([]) == 0

# test case 3: one element
def test_one_element():
    assert sum_of_list([10]) == 10
    assert sum_of_list([0]) ==0
```

```python
import pytest

def factorial(n):
    if n == 0: return 1
    elif (type(n) == int):
        return n * factorial(n-1)
    else: return -1

# Test case: expected input
def test_regular():
    assert factorial(5) == 120

# Test case: zero input
def test_zero():
    assert factorial(0) == 1

# Test case: input of a wrong type
def test_str():
    assert factorial('5') == -1
```

## Feature test
Una característica (feature) es una funcionalidad del sistema desoftware que satisface los requisitos de un usuario particular.

Una feature es una parte de software que es más amplia que una unidad. Pero a veces podemos implementar una feature como una unidad, por lo que coincidirán completamente.
Por lo tanto una feature testing es un método de prueba que permite verificar el comportamiento de una feature específica.
